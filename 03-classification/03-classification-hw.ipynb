# Question 5: feature elimination technique

```python
#Train a model using the same features and parameters as in Q4 (without rounding).
#Now exclude each feature from this set and train a model without it. Record the accuracy for each model.
#For each feature, calculate the difference between the original accuracy and the accuracy without the feature.

from sklearn.feature_extraction import DictVectorizer
def vectorize_data(df):
    dv = DictVectorizer(sparse=False)
    return dv.fit(df.to_dict(orient='records'))

from sklearn.linear_model import LogisticRegression
def fit_model(X_train, y_train):
    model = LogisticRegression(solver='liblinear', C=1.0, max_iter=1000, random_state=42)
    model.fit(X_train, y_train)
    return model

def calculate_accuracy(y_pred, y_val):    
    df_pred = pd.DataFrame()
    df_pred['probability'] = y_pred
    df_pred['prediction'] = (y_pred >= 0.5).astype(int)
    df_pred['actual'] = y_val
    df_pred['correct'] = df_pred.prediction == df_pred.actual
    return df_pred.correct.mean()


def train(df_train, y_train):
    vectorizer = vectorize_data(df_train)
    X_train = vectorizer.transform(df_train.to_dict(orient='records'))
    model = fit_model(X_train, y_train)
    return model, vectorizer

def validate_model(model_description, model, X_val, y_val):
    accuracy = calculate_accuracy(model.predict_proba(X_val)[:, 1], y_val)
    df_accuracy= pd.DataFrame([{"Features": model_description, "accuracy": accuracy}])
    return df_accuracy

def train_and_validate_model(model_description, df_train, y_train, df_val, y_val):
    (model, vectorizer) = train(df_train, y_train)
    X_val = vectorizer.transform(df_val.to_dict(orient='records'))
    df_accuracy = validate_model(model_description, model, X_val, y_val)
    return df_accuracy

training_comparison = pd.DataFrame()
df_accuracy = train_and_validate_model("Alls", df_train, y_train, df_val, y_val)
initial_accuracy = df_accuracy['accuracy'].values[0]
df_accuracy['accuracy_improvement'] = df_accuracy['accuracy']-initial_accuracy
training_comparison = pd.concat([training_comparison, df_accuracy], ignore_index=True)


for c in df_train.columns:
    df_train_c = df_train.copy()
    df_train_c.drop(columns=[c], inplace=True)
    df_val_c = df_val.copy()
    df_val_c.drop(columns=[c], inplace=True)
    df_accuracy = train_and_validate_model(f"Alls - {c}", df_train_c, y_train, df_val_c, y_val)
    df_accuracy['accuracy_improvement'] = df_accuracy['accuracy']-initial_accuracy
    training_comparison = pd.concat([training_comparison, df_accuracy], ignore_index=True)

training_comparison
```

#  Question 6: Parameter tuning

```python
# Let's try the following values of the parameter C: [0.000001, 0.00001, 0.0001, 0.001].
# Train models using all the features as in Q4.
# Calculate the accuracy on the validation dataset and round it to 3 decimal digits.

def fit_model(C, X_train, y_train):
    model = LogisticRegression(solver='liblinear', C=C, max_iter=1000, random_state=42)
    model.fit(X_train, y_train)
    return model

def train(C, df_train, y_train):
    vectorizer = vectorize_data(df_train)
    X_train = vectorizer.transform(df_train.to_dict(orient='records'))
    model = fit_model(C, X_train, y_train)
    return model, vectorizer

def train_and_validate_model(model_description, C, df_train, y_train, df_val, y_val):
    (model, vectorizer) = train(C, df_train, y_train)
    X_val = vectorizer.transform(df_val.to_dict(orient='records'))
    df_accuracy = validate_model(model_description, model, X_val, y_val)
    return df_accuracy

training_comparison = pd.DataFrame()

C_values = [0.000001, 0.00001, 0.0001, 0.001]
for C in C_values:
    df_accuracy = train_and_validate_model(f"{C}", C, df_train, y_train, df_val, y_val)
    training_comparison = pd.concat([training_comparison, df_accuracy], ignore_index=True)

training_comparison
```
