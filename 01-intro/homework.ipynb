Q6. Median value of horsepower

```python
horsepower_median = df["horsepower"].median()
print(f"The median horsepower: {horsepower_median}")

horsepower_frequency = df.groupby("horsepower")["horsepower"].count().sort_values(ascending=False)
#type(horsepower_frequency)   # <class 'pandas.core.series.Series'>
horsepower_most_frequent = horsepower_frequency.head(1)
horsepower_most_frequent_value = horsepower_most_frequent.index[0]
print(f"The most frequent horsepower: {horsepower_most_frequent_value}")

print(f"Built in function to calculate the most frequent horsepower: {df["horsepower"].mode()[0]}")
```

```python
#fill the missing values in the horsepower column with the most frequent value
df_horsepower_modified = df["horsepower"].fillna(horsepower_most_frequent_value)
print(f"New median after removing the missing values: {df_horsepower_modified.median()}")
```

Q7. Sum of weights

```python
#Select all the cars from Asia
#Select only columns vehicle_weight and model_year
#Select the first 7 values
#Get the underlying NumPy array. Let's call it X.
X = df[df["origin"] == "Asia"][["vehicle_weight","model_year"]].head(7).values
X
```

```python
# Get the transpose of X:
X.T
```

```python
# Compute matrix-matrix multiplication between the transpose of X and X.
#XTX = X.T @ X
XTX = X.T.dot(X)
XTX
```

```python
import numpy as np
# Compute the inverse of XTX.
XTX_inverse = np.linalg.inv(XTX)
XTX_inverse
```

```python
#Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
#Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w = XTX_inverse.dot(X.T).dot(y)
w
```

```python
w.sum()
```
